````python
import sys
sys.path.append('../src')

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style='whitegrid')
%matplotlib inline
````

## Part 1: Mall Customer Segmentation

````python
mall_df = pd.read_csv('../data/Mall_Customers.csv')
mall_df.head()
````

````python
mall_df.describe()
````

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(mall_df['Age'], kde=True, ax=axes[0])
sns.histplot(mall_df['Annual Income (k$)'], kde=True, ax=axes[1])
sns.histplot(mall_df['Spending Score (1-100)'], kde=True, ax=axes[2])
plt.tight_layout()
````

````python
# Reuse the pipeline functions from src/mall_segmentation.py
from mall_segmentation import select_features, find_optimal_k, fit_kmeans, profile_clusters
from sklearn.preprocessing import StandardScaler

feature_cols = ['Annual Income (k$)', 'Spending Score (1-100)']
X = select_features(mall_df, feature_cols)
X_scaled = StandardScaler().fit_transform(X)

best_k, inertias, sil_scores = find_optimal_k(X_scaled)
model = fit_kmeans(X_scaled, best_k)
labeled_df = profile_clusters(mall_df, feature_cols, model.labels_)
````

## Part 2: Big Mart Sales Prediction
